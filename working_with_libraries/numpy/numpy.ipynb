### Vectors and such



A vector can be thought of as an ordered list of numbers (in NumPy, arrays can also hold other uniform types such as integers, floats, or booleans). In mathematics, vectors are used to represent points or directions in space, whether in two dimensions, three dimensions, or beyond. Linear algebra employs vector notation to efficiently illustrate geometric relationships and systems of linear equations, while also offering numerous methods for manipulating vector and matrix data, such as the dot product, cross product, matrix inversion, eigenvectors etc. Indeed, vector arithmetic forms the foundation of most computational techniques, playing a crucial role in nearly all artificial intelligence endeavors. The meteoric rise of Nvidia is owed to the fact that GPUs are particularly well suited for vector-based computations. Let&rsquo;s take a look at a simple 2-D vector addition:

Let $\mathbf{a} = \begin{bmatrix} 2 \\ 3 \end{bmatrix}$ and $\mathbf{b} = \begin{bmatrix} 4 \\ 1 \end{bmatrix}$. 

The addition of these two vectors is given by:

$$
\mathbf{c} = \mathbf{a} + \mathbf{b} = \begin{bmatrix} 2 \\ 3 \end{bmatrix} + \begin{bmatrix} 4 \\ 1 \end{bmatrix} = \begin{bmatrix} 2 + 4 \\ 3 + 1 \end{bmatrix} = \begin{bmatrix} 6 \\ 4 \end{bmatrix}
$$

You will notice that the vector elements are added element wise. If you try this with a list, it will not work



```python
a = [2, 3]
b = [4, 1]
c = a + b
print(c)
```

#### Solving a system of linear equations



Imagine you meet a friend for coffee. You buy 2 coffees and a muffin and you pay $10. Your friend buys one coffee and 3 muffins and pays $15. Afterward you try to deduce the cost of a muffin and the cost of a coffee. Mathematically, you can express this as two equations

\begin{align}
2C + 1 M  &= 10 \\
1C + 3 M &= 15
\end{align}

Here, we have two unknowns (`C` and `M`) and two independent equations, so this problem has a unique solution.

Solving a system of linear equations is a classical linear algebra problem that is conveniently expressed in matrix and vector form, where you place the quantities of coffee and muffins into the matrix and the amount of monies paid into a vector. We can then write:

\begin{equation}
\begin{bmatrix}
2 & 1\\
1 & 3\\
\end{bmatrix}  
\begin{bmatrix} C\\ M\\ \end{bmatrix}  = \begin{bmatrix}10\\ 15\\ \end{bmatrix}
\end{equation}

And this equation is then solved for the unknown `C` & `M` like this.

\begin{equation}
\begin{bmatrix}
 C\\ M\\ \end{bmatrix}  = \begin{bmatrix} 2 & 1\\ 1 & 3\\ \end{bmatrix}^{-1}  \cdot \begin{bmatrix}10\\ 15\\ \end{bmatrix}
\end{equation}

Where the negative exponent denotes the inverse of the matrix. Using NumPy, we can solve the above like this:



```python
import numpy as np

# The "Menu" (Quantities of Coffee and Muffins)
menu = np.array([
    [2, 1], 
    [1, 3]
])

# The "Receipts" (Total cost of each order)
totals = np.array([10, 15])

# Ask Numpy to find the individual prices through matrix inversion.
# Note: in practice, np.linalg.solve(menu, totals) is preferred because
# it is faster and more numerically accurate than computing the inverse.
prices = np.linalg.inv(menu) @ totals

print(f"Price of one Coffee: ${prices[0]:.2f}")
print(f"Price of one Muffin: ${prices[1]:.2f}")
```

Note that in linear algebra there are different ways to multiply vectors and matrices, such as matrix multiplication or  the dot product (`@`, `np.matmul`, `np.dot`) and the cross product (`np.cross`)  - see [https://betterexplained.com/articles/cross-product/](https://betterexplained.com/articles/cross-product/)



#### Plotting vectors as arrows



2-D vector math is easily visualized with the Matplotlib arrow  object.  In the code below I use the `*` and `**` operators to achieve more compact and readable form. You may recall that Python automatically expands list type objects when we write `a, b = [1,3]`. There are however cases where this does not happen automatically, e.g., if you pass a list to a function. We can use the `*` to manually unpack list-type data and the `**` operator to unpack dictionaries . This way we do have to repeat lengthy default options each time we plot an arrow.



```python
import numpy as np
import matplotlib.pyplot as plt

# ---- define vectors ---------------- #
origin = (0,0)  # lets assume that all vectors start at 0,0
v = np.array([1,2])
u = np.array([1,-3])

# ---- vector math ------------------- #

# ---- set default arrow options ----- #
arrow_options = {"length_includes_head":True,
                 "width":0.01,
                 "head_width":0.1,
                 }

# ---- plot vectors ------------------ #
fig, ax = plt.subplots()
ax.arrow(*origin, *v,color="C0",label="v",**arrow_options)
ax.arrow(*origin, *u,color="C1",label="u",**arrow_options)
ax.legend()
plt.show()
```

#### Rotating a vector with a rotation matrix



Vector addition allows us to translate vectors in space. If we want to rotate a vector, we can use a rotation matrix like this:

\begin{equation}
\begin{bmatrix} x' \\ y' \end{bmatrix} = \begin{bmatrix} \cos(\theta) & -\sin(\theta) \\ \sin(\theta) & \cos(\theta) \end{bmatrix} \begin{bmatrix} x \\ y \end{bmatrix}
\end{equation}

Let&rsquo;s do this with Python



```python
import numpy as np
from math import sin, cos, radians
import matplotlib.pyplot as plt

# ---- define vectors ---------------- #
origin = (0,0)  # lets assume that all vectors start at 0,0
u = np.array([1,0]) # set vector
phi = radians(30) # sin/cos expect radians = deg * pi/180
M = [[cos(phi), -sin(phi)],
     [sin(phi), cos(phi)],
     ]

# multiply matrix by vector
v = M @ u # Matrix first, vector last!

# ---- set default arrow options ----- #
arrow_options = {"length_includes_head":True,
                 "width":0.01,
                 "head_width":0.1,
                 }

# ---- plot vectors ------------------ #
fig, ax = plt.subplots()
ax.arrow(*origin, *u,color="C0",label="u",**arrow_options)
ax.arrow(*origin, *v,color="C1",label="v",**arrow_options)
ax.legend()
plt.show()
```

We can extend this notation for 3 dimensions (here rotating around the $z$-axis) and add the translation amount into the matrix as well:

\begin{equation}
\begin{bmatrix} 
x' \\ y' \\ z' \\ 1 
\end{bmatrix} = \begin{bmatrix} 
\cos(\theta) & -\sin(\theta) & 0 & t_x \\ 
\sin(\theta) & \cos(\theta) & 0 & t_y \\ 
0 & 0 & 1 & t_z \\ 
0 & 0 & 0 & 1 
\end{bmatrix} \begin{bmatrix} x \\ y \\ z \\ 1 \end{bmatrix}
\end{equation}

What you end up with is the foundation of all computer graphics (mapping projections etc.). And before you ask, the additional 1 is required because translation cannot be expressed by a 3 by 3 matrix alone - by adding an extra row and column (making it a 4 by 4 matrix) and appending a 1 to the vector, we can combine rotation and translation into a single matrix multiplication. This notation is also called homogeneous coordinates. 

Let&rsquo;s try this in action:



```python
"""Example showing the use of homogeneous coordinates to rotate and
translate coordinates.

Notes:
  - vectors need to be of size 4, with x, y, z, w
  - w must be set to 1 so that translation is applied normally (w=0 disables translation)
"""

import numpy as np
from math import sin, cos, radians
import matplotlib.pyplot as plt

# ---- define vectors ---------------- #
origin = np.array([0, 0, 0, 1])  # lets assume that all vectors start at 0,0
u = np.array([1, 0, 0, 1])  # set vector end point

# set translation offset
tx, ty, tz = [1, 2, 0]

# set rotation amount in the x/y plane
phi = radians(30)  # sin/cos expect radians = deg * pi/180

# define transformation matrix
M = [
    [cos(phi), -sin(phi), 0, tx],
    [sin(phi), cos(phi), 0, ty],
    [0, 0, 1, tz],
    [0, 0, 0, 1],
]

# transform the origin and end points for each vector
origin_t = M @ origin  # apply transformation matrix
v = M @ u  # Matrix first, vector last!

# ---- set default arrow options ----- #
arrow_options = {
    "length_includes_head": True,
    "width": 0.01,
    "head_width": 0.1,
}

# ---- plot vectors ------------------ #
fig, ax = plt.subplots()
# plot original vector
ax.arrow(*origin[:2], *u[:2], color="C0", label="u", **arrow_options)

# plot transformed vector
ax.arrow(*origin_t[:2], *(v[:2] - origin_t[:2]), color="C1", label="v", **arrow_options)
ax.legend()
plt.show()
```

### Linear Interpolation



NumPy offers a wide array of powerful functions for various mathematical computations. One notable example is its interpolation function. Linear interpolation is a technique used to estimate values within the range of a set of known data points. This method is particularly useful when dealing with discrete data and needing to estimate an intermediate value between two observed points. Essentially, linear interpolation assumes that the change between these two points is linear, allowing it to be approximated by a straight line.

Mathematically, if we have two known points $(x_1, y_1)$ and $(x_2, y_2)$, and we want to estimate the value of $y$ for an intermediate value $x$, we use the following formula for linear interpolation:

$$
y = y_1 + \frac{(x - x_1)}{(x_2 - x_1)} \cdot (y_2 - y_1)
$$

Here:

-   $x_1$ and $x_2$ are the known $x$-values (data points),
-   $y_1$ and $y_2$ are the corresponding $y$-values (data points),
-   $x$ is the value where we would like to estimate $y$.

This formula essentially calculates how far $x$ is between $x_1$ and $x_2$ (the weight or fraction) and applies the same proportion to the $y$ values. Linear interpolation is widely used in cases where a quick and straightforward method is needed to fill gaps or generate continuous transitions in data.



#### A practical example



Below, we demonstrate how to use NumPy&rsquo;s `np.interp` function for linear interpolation. We&rsquo;ll create a simple dataset of known $x$-values and their corresponding $y$-values, and then estimate the $y$-values for a set of new $x$-values not present in the original data. We&rsquo;ll also visualize the results using Matplotlib to show how interpolation works.



```python
import numpy as np
import matplotlib.pyplot as plt

# Known data points
x_known = np.array([1, 2, 3, 4, 5])  # x-values
y_known = np.array([2, 4, 6, 8, 10])  # y-values

# New x-values (where we want to estimate y-values)
x_new = np.array([1.5, 2.5, 3.5, 4.5])

# Perform linear interpolation using np.interp
y_new = np.interp(x_new, x_known, y_known)

# Plot the results
plt.figure(figsize=(8, 5))

# Plot the known data points
plt.scatter(x_known, y_known, color='blue', label='Known Points', zorder=3)
plt.plot(x_known, y_known, color='blue', linestyle='--', alpha=0.6, label='Known Line')

# Plot the interpolated points
plt.scatter(x_new, y_new, color='red', label='Interpolated Points', zorder=3)
# Add titles and labels
plt.show()
```

#### Explanation of the Code



1.  Input Data: We provide `x_known` and `y_known` arrays that define the known data points.
2.  New Points: The `x_new` array contains the $x$ values where we want to estimate $y$ values.
3.  Linear Interpolation: The `np.interp` function takes three arguments:
    -   The new $x$ values (`x_new`) are where $y$ values need to be interpolated.
    -   The original $x$ values (`x_known`).
    -   The original $y$ values (`y_known`).
4.  Output: It returns the interpolated $y$ values (`y_new`) corresponding to the $x_{\text{new}}$ values.
5.  Visualization: Known data points are visualized with blue dots, and interpolated points are shown in red. The dashed blue line offers visual guidance for the known trend.



#### Sample Outcome



The figure will show:

-   The known points as blue circles.
-   The interpolated points are as red dots at fractional positions (e.g., $x = 1.5, y = 3$).
-   A dotted blue line connecting the data, illustrating the linear relationships.
